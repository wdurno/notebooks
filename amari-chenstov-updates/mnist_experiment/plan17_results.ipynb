## Settings And Glossary

**Action pi** weights the arriving batch in both diagonal Fisher forgetting and EWC odds; **prevalence p** controls digit-9 arrivals, not the action. The action is frozen before the current batch; its observed-label score updates the observer for the next action only. **Anchor** is a common local score coordinate, not an oracle. The **observation clock** advances on incoming batches, including quiet holds.

Observer scores differentiate log likelihood using the stream's observed labels; Fisher diagonals integrate over model labels instead. These are not interchangeable.

**Movement** is the observer's posterior mean contribution; **uncertainty** is a separate covariance contribution, not proof of environmental motion. Kalman covariance is not another Fisher estimand. **Covariance-only** sets movement energy to zero in the action rule; it is not the Kalman filter's covariance. EMA is a Holt level/trend point plug-in with possible squared-noise bias. Kalman versus EMA tests the complete estimator package, including uncertainty, not just its gain.

**9 recall** conditions on true 9s; **old retention** is macro accuracy over digits 0--8. Precision is standardized to fixed p_ref=.1 and remains undefined with no predicted positives. Predictive metrics use independent stratified holdouts, separate from training and Fisher inputs; development and fresh evaluation holdouts are independent. Predictive NLL is pure NLL, excluding every regularizer. Mixture-weighted all-accuracy is secondary, never a success gate.

A step is one incoming batch; `arrivals = step * samples_per_step` counts **online digits observed**, excluding initialization. Exact nominal concentration q gives weight-based effective size 1/q. The product of (1-pi) is the initialization's nominal remaining weight, not measured knowledge.

### Phase 7 Conclusion: Adaptation When Tuning Trials Are Unavailable

**Velocity Kalman is a promising way to choose adaptation weights from one continuing data stream when repeated tuning trials are unavailable.** Once configured, it uses past score observations to estimate movement and noise and chooses each weight before the next batch. Deployment needs neither competing learner runs nor holdout feedback. This fits continual learning with limited data and compute, where finding a good fixed weight through repeated training may be impractical.

Phase 7 provides encouraging evidence: modest current-distribution tracking gains over fixed .025, including both ramp directions, with slightly lower measured online wall time. Automatically choosing a useful average weight could itself be valuable; the benefit need not come entirely from precisely timed changes.

The observer settings were selected on development trajectories, so this is a calibrated recipe whose transfer to unfamiliar streams remains to be tested. The present comparisons do not separate average-weight selection from adaptive timing, and fixed endpoint prediction worsened. The practical opportunity is to reduce dependence on unavailable tuning trials while continuing to measure both tracking and retention.
