```python
# I'm addicted to markdown in notebooks, so we'll run everything from here. 
# Run torch.distributed and FSDP as subprocess scripts because they weren't made to work with IPython. 

import sys
import subprocess

def run_script(script_path, *args):
    """
    Runs a Python script as a subprocess using the current Jupyter kernel's Python interpreter.
    
    Args:
        script_path (str): Path to the Python script to run.
        *args: Additional arguments to pass to the script.

    Returns:
        int: Exit code of the script.
    """
    python_executable = sys.executable  # Get the current kernel's Python interpreter

    command = [python_executable, script_path] + list(map(str, args))  # Build command
    print(f"Running script: {script_path}\nUsing Python: {python_executable}\nCommand: {' '.join(command)}")

    try:
        # Run the subprocess
        result = subprocess.run(command, text=True, capture_output=True)

        # Print stdout and stderr in notebook
        print("\n--- Script Output ---")
        if result.stdout:
            print(result.stdout)
        
        if result.stderr:
            print("\n--- Script Errors ---")
            print(result.stderr)

        return result.returncode  # Return the exit code

    except Exception as e:
        print(f"\nError running script {script_path}: {e}")
        return -1  # Return a failure code
```

```python
run_script('torch_dist_example.py')
```
